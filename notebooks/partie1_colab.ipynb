```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

On prépare les données (mes résultats dans sql)

- Répartion des clients pas pays



```python
data = {
    "country": ["USA", "France", "Germany", "Japan"],
    "nb_clients": [24, 12, 10, 8]
}

df = pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="country", y="nb_clients", palette="viridis")
plt.title("Répartition des clients par pays")
plt.xlabel("Pays")
plt.ylabel("Nombre de clients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

- Répartition des commandes par status

```python
data = {
    "status": ["Shipped", "Cancelled", "On Hold", "Resolved", "Disputed"],
    "nb_orders": [250, 12, 30, 18, 5]
}

df = pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="status", y="nb_orders", palette="magma")
plt.title("Répartition des commandes par statut")
plt.xlabel("Statut")
plt.ylabel("Nombre de commandes")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

- Répresentation des 10 produits les plus vendus

```python
data = {
    "productName": [
        "Produit A", "Produit B", "Produit C", "Produit D", "Produit E",
        "Produit F", "Produit G", "Produit H", "Produit I", "Produit J"
    ],
    "total_sold": [1200, 1100, 950, 900, 850, 800, 780, 760, 740, 700]
}

df_top10 = pd.DataFrame(data)
df_top10
```

```python
plt.figure(figsize=(14, 6))
sns.barplot(data=df_top10, x="productName", y="total_sold", palette="viridis")
plt.title("Top 10 des produits les plus vendus")
plt.xlabel("Produit")
plt.ylabel("Quantité vendue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

- Répresentation des 5
meilleurs clients (par chiffre d'affaires)

```python
data = {
    "customerName": ["Client A", "Client B", "Client C", "Client D", "Client E"],
    "revenue": [150000, 120000, 110000, 95000, 90000]
}

df_top5_clients = pd.DataFrame(data)
df_top5_clients
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df_top5_clients, x="customerName", y="revenue", palette="magma")
plt.title("Top 5 des meilleurs clients (par chiffre d'affaires)")
plt.xlabel("Client")
plt.ylabel("Chiffre d'affaires (€)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

- Répresentation de la gamme de produits la plus rentable

```python
data = {
    "productLine": [
        "Classic Cars", "Motorcycles", "Planes",
        "Ships", "Trains", "Vintage Cars"
    ],
    "revenue": [450000, 320000, 280000, 260000, 200000, 180000]
}

df_productline = pd.DataFrame(data)
df_productline
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df_productline, x="productLine", y="revenue", palette="coolwarm")
plt.title("Gamme de produits la plus rentable")
plt.xlabel("Gamme de produits")
plt.ylabel("Chiffre d'affaires (€)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
